# SQL Server Sample Queries

This notebook demonstrates SQL queries using the SQL Notebook Kernel.

## Prerequisites

1. Configure connection string in VS Code settings:
   ```json
   {
     "sqlNotebook.defaultConnectionString": "Server=localhost;Database=AdventureWorks;User Id=sa;Password=YourPassword;Encrypt=true;TrustServerCertificate=true"
   }
   ```

2. Select **SQL Kernel** from the kernel picker (top-right)

````sql
-- Simple SELECT query
SELECT 
    DB_NAME() AS DatabaseName,
    @@VERSION AS SQLServerVersion,
    GETDATE() AS CurrentDateTime;
````

````sql
-- Create a temporary table
CREATE TABLE #TempCustomers (
    CustomerID INT PRIMARY KEY,
    CustomerName NVARCHAR(100),
    City NVARCHAR(50),
    Country NVARCHAR(50)
);

INSERT INTO #TempCustomers VALUES
(1, 'John Doe', 'New York', 'USA'),
(2, 'Jane Smith', 'London', 'UK'),
(3, 'Bob Johnson', 'Toronto', 'Canada'),
(4, 'Alice Williams', 'Sydney', 'Australia'),
(5, 'Charlie Brown', 'Berlin', 'Germany');

SELECT * FROM #TempCustomers;
````

````sql
-- Multiple statements with GO
SELECT COUNT(*) AS TotalCustomers FROM #TempCustomers;
GO

SELECT Country, COUNT(*) AS CustomerCount
FROM #TempCustomers
GROUP BY Country
ORDER BY CustomerCount DESC;
GO
````

````sql
-- Update and show results
UPDATE #TempCustomers 
SET City = 'Los Angeles' 
WHERE CustomerID = 1;

SELECT * FROM #TempCustomers WHERE CustomerID = 1;
````

````sql
-- Clean up
DROP TABLE #TempCustomers;
````
